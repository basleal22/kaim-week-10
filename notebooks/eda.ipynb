```python
data= pd.read_csv('data/BrentOilPrices.csv')
```

```python
data
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(data.index, data["Price"], label="Brent Oil Price", color="blue")
plt.xlabel("Year")
plt.ylabel("Price (USD)")
plt.title("Brent Oil Prices Over Time")
plt.legend()
plt.show()
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose

decomposition = seasonal_decompose(data["Price"], model="additive", period=365)  # Assuming daily data
decomposition.plot()
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

adf_test = adfuller(data["Price"])
print(f"ADF Statistic: {adf_test[0]}")
print(f"P-value: {adf_test[1]}")
```

p-value > 0.05 meaning Data is non-stationary (need differencing)

now we assess for the Price Volatility (Moving Standard Deviation)

```python
data["Rolling_Std"] = data["Price"].rolling(window=30).std()

plt.figure(figsize=(12,6))
plt.plot(data.index, data["Rolling_Std"], label="30-Day Rolling Std Dev", color="green")
plt.xlabel("Year")
plt.ylabel("Price Volatility")
plt.title("Brent Oil Price Volatility")
plt.legend()
plt.show()
```
