```python
oil_data=pd.read_csv('data/BrentOilPrices.csv')
wb_gdp=pd.read_csv('data/wb_gdp.csv')
wb_gdp=wb_gdp.rename(columns={'date':'Date '})
wb_gdp
```

```python
wb_gdp=wb_gdp.drop(columns='Unnamed: 0')
```

```python
print(wb_gdp.columns)
print(oil_data.columns)
wb_gdp.rename(columns={'Date ': 'Date'}, inplace=True)
#oil_data.rename(columns={'Date ': 'Date'}, inplace=True)
```

```python
# Convert 'Date' from year to datetime (assuming it's stored as a numeric type)
wb_gdp['Date'] = pd.to_datetime(wb_gdp['Date'].astype(str) + '-01-01')

# Now extract the year
wb_gdp['Year'] = wb_gdp['Date'].dt.year
# Extract the year from 'Date' in oil_data
# Convert 'Date' to datetime in oil_data
oil_data['Date'] = pd.to_datetime(oil_data['Date'], errors='coerce')
# Extract year from the Date column in oil_data
oil_data['Year'] = oil_data['Date'].dt.year
# Verify the results
print("Unique years in wb_gdp:", wb_gdp['Year'].unique())
# Group oil_data by year and take the mean for each year
oil_data_grouped = oil_data.groupby('Year').agg({'Price': 'mean'}).reset_index()

# Now merge with wb_gdp
merged_data = pd.merge(wb_gdp, oil_data_grouped, on='Year', how='inner')
print(merged_data.shape)
```

```python
merged_data.to_csv('merged_data.csv')
```
